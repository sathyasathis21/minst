## problem statement :
build a Machine learning model to accurately predict whether or not the patients in the datasets have diabets or not.

## Import necessary libraries

```python
import pandas as pd
import keras
import warnings
warnings.filterwarnings("ignore")
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from keras.layers import Dense,Dropout
from keras.models import Sequential
from keras.models import load_model
```

## import data

```python
diabetics_data=pd.read_csv("diabetes.csv")
diabetics_data
```

## Data understanding

```python
diabetics_data.isnull().sum()
```

```python
diabetics_data.dtypes
```

```python
diabetics_data.shape
```

## Data preprocessing

```python
x=diabetics_data.drop("Outcome",axis=1)
y=diabetics_data["Outcome"]
```

```python
x
```

```python
y
```

## Model building

```python
model=Sequential()
model.add(Dense(units=13,input_dim=8,kernel_initializer="uniform",activation="relu"))
model.add(Dense(units=8,kernel_initializer="uniform",activation="relu"))
model.add(Dense(units=1,kernel_initializer="uniform",activation="sigmoid"))
#model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"])
```

```python
 x_train, x_test, y_train, y_test =train_test_split(x,y,test_size=0.20,random_state=123,shuffle=True,stratify=y)
```

```python
x_test.shape,y_test.shape
```

```python
x_train.shape,y_train.shape
```

### Alert:- Here's where you need to apply standardization fit_transform,transform

```python
scalar=StandardScaler()
x_train=scalar.fit_transform(x_train)
x_test=scalar.transform(x_test)
```

## MODEL TRAINING || MODEL TESTING || MODEL EVALUATION

```python
model.fit(  x=x_train,
    y=y_train,
    batch_size=16,
    epochs=50,validation_split=0.2)
```

## Model prediction

```python
y_pred=model.predict(x_test)
y_pred=(y_pred>0.60).astype(int)
y_pred
```

```python
model.save("diabetics_data.h5")
```

### Bonus :- here you can check other metric,for structured data

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```
